## The data
### **netflix_data.csv**
| Column | Description |
|--------|-------------|
| `show_id` | The ID of the show |
| `type` | Type of show |
| `title` | Title of the show |
| `director` | Director of the show |
| `cast` | Cast of the show |
| `country` | Country of origin |
| `date_added` | Date added to Netflix |
| `release_year` | Year of Netflix release |
| `duration` | Duration of the show in minutes |
| `description` | Description of the show |
| `genre` | Show genre |

```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

```python
# filtering my dataset 
df = netflix_df[(netflix_df['type'] == 'Movie') & (netflix_df['release_year'] >= 1990) & (netflix_df['release_year'] <= 1999)]
```

```python
# our new subset 
df
```

```python
# creating a histogram showing the distribution of movie durations from 1990-1999 and then calculating the mode

plt.figure(figsize = (12,6))
plt.hist(df['duration'], bins = 25, color = 'teal', edgecolor = 'black', alpha = 0.5)
plt.title('Distribution of Movie Durations between 1990-1999')
plt.xlabel('minutes')
plt.ylabel('frequency')
plt.show()

duration_mode = df['duration'].mode()
duration = int(duration_mode)
duration
```

```python
# creating my second sub set 

df_2 = netflix_df[(netflix_df['type'] == 'Movie') & (netflix_df['release_year'] >= 1990) & (netflix_df['genre'] == 'Action') & (netflix_df['release_year'] <= 1999)]

df_2
```

```python
# counting the number of movies with a duration of less 90 mins 

short_movie_count = 0

for label, row in df_2.iterrows():
    if row['duration'] < 90:
        short_movie_count += 1

short_movie_count
    
```
